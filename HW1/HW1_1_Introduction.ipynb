## 2. Assigning variables
Gibbs free energy $\Delta G = \Delta H - T\Delta S$, with the values used in the lesson.

```python
deltaH = -541.5   # kJ/mol
deltaS = 10.4     # kJ/(mol K)
temp = 298        # K
deltaG = deltaH - temp * deltaS
```

```python
print(deltaG)
```

An expression that is not assigned to a variable does not change the variable:

```python
print(deltaG)
deltaG * 1000
print(deltaG)
```

To change it, the result must be assigned again (overwriting the variable):

```python
print(deltaG)
deltaG = deltaG * 1000
print(deltaG)
```

Usually it is better practice to store the result under a new, informative name:

```python
print(deltaG)
deltaG_joules = deltaG * 1000
print(deltaG)
print(deltaG_joules)
```

## 5. Lists

```python
# This is a list
energy_kcal = [-13.4, -2.7, 5.4, 42.1]
# Its length
energy_length = len(energy_kcal)
print('The length of this list is', energy_length)
```

```python
# First element (Python counts from 0)
print(energy_kcal[0])
```

```python
# Convert the second element to kJ
energy_kilojoules = energy_kcal[1] * 4.184
print(energy_kilojoules)
```

## 6. Slices
`list[start:end]` includes `start` but **not** `end`.

```python
short_list = energy_kcal[0:2]
print(short_list)
```

**Check your understanding:** what does the following code print?

```python
slice1 = energy_kcal[1:]
slice2 = energy_kcal[:3]
print('slice1 is', slice1)
print('slice2 is', slice2)
```

Taking a slice without assigning it does not modify the original list:

```python
print(energy_kcal)
energy_kcal[0:2]
print(energy_kcal)
```

## 7. `for` loops

```python
for number in energy_kcal:
    kJ = number * 4.184
    print(kJ)
```

Using `append` on a list that has not been created yet fails with a `NameError`.
(The error is caught here so the notebook can run from top to bottom.)

```python
try:
    for number in energy_kcal:
        kJ = number * 4.184
        energy_kJ.append(kJ)
except NameError as err:
    print('NameError:', err)
```

The list must be created (empty) **before** the loop:

```python
energy_kJ = []
for number in energy_kcal:
    kJ = number * 4.184
    energy_kJ.append(kJ)

print(energy_kJ)
```

## 8. Logic statements (`if`)

```python
negative_energy_kJ = []

for number in energy_kJ:
    if number < 0:
        negative_energy_kJ.append(number)

print(negative_energy_kJ)
```

```python
negative_numbers = []
for number in energy_kJ:
    if number < 0 or number == 0:
        negative_numbers.append(number)

print(negative_numbers)
```

## 9. Exercise
Loop over `data_list`; if an element is a string, recast it as a float. Save *all* numbers in `number_list`.

```python
data_list = ['-12.5', 14.4, 8.1, '42']
number_list = []
for item in data_list:
    if type(item) is str:
        item = float(item)
    number_list.append(item)

print(number_list)
```

```python
# Check: every element is now a float
print([type(x) for x in number_list])
```
